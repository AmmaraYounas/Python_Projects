
Jane's favorite book has a secret message hidden in its title: "The Sceret Sotry of Pyhton Cdoe". Print the title with the correct letter order by swapping adjacent characters.


```python

title = "The Sceret Sotry of Pyhton Cdoe"
corrected_title = "".join(title[i:i+2][::-1] for i in range(0, len(title), 2))
print(corrected_title)
```


James found a string containing a secret code: "1J2a3m4e5s". Extract and print only the alphabetical characters from the string.


```python

secret_code = "1J2a3m4e5s"
alphabetical_chars = "".join(char for char in secret_code if char.isalpha())
print(alphabetical_chars)
```
